```python
import pandas as pd
```

```python
from sqlalchemy import create_engine
```

```python
from time import time
```

```python
# engine = create_engine('postgresql://usuario:contraseña@localhost:5432/nombre_base_datos')
```

```python
engine = create_engine('postgresql://root:root@localhost:5432/ny_taxi')
```

```python
try:
    with engine.connect() as connection:
        # El comando TRUNCATE TABLE es un comando de SQL
        connection.execute("TRUNCATE TABLE yellow_taxi_data;")
        # Si la conexión se maneja automáticamente con un contexto (como engine.connect()), 
        # se aplicará automáticamente al salir del bloque 'with'.
    print("Tabla truncada exitosamente.")

except Exception as e:
    print(f"Ocurrió un error al truncar la tabla: {e}")
                           
```

```python
df_iter = pd.read_csv('nyt.csv', iterator=True, chunksize=100000)
```

```python
try:
    while True:
        t_start = time()

        try:
            df = next(df_iter)
        except StopIteration:
            print('No more data to process')
            break
            
        df.tpep_pickup_datetime = pd.to_datetime(df.tpep_pickup_datetime)
        df.tpep_dropoff_datetime = pd.to_datetime(df.tpep_dropoff_datetime)

        try:
            df.to_sql(name='yellow_taxi_data', con=engine, if_exists='append')
        except Exception as e:
            print(f'Error inserting chunk into database: {e}')
            continue
        t_end = time()

        print(f'Inserted another chunk, took {t_end - t_start} seconds')


except Exception as e:
    print(f'An error occurred during processing: {e}')
```
