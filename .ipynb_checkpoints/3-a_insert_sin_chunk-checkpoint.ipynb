```python
import pandas as pd
```

```python
from sqlalchemy import create_engine
```

```python
from time import time
```

```python
# engine = create_engine('postgresql://usuario:contraseña@localhost:5432/nombre_base_datos')
```

```python
engine = create_engine('postgresql://root:root@localhost:5432/ny_taxi')
```

```python
df1 = pd.read_csv('nyt.csv')
```

```python
df1.tpep_pickup_datetime = pd.to_datetime(df1.tpep_pickup_datetime)
df1.tpep_dropoff_datetime = pd.to_datetime(df1.tpep_dropoff_datetime)
```

Creamos una tabla en la BBDD. Se llama yellow_taxi_data y si existe la sobreescribe

```python
df1.head(n=0).to_sql(name='yellow_taxi_data', con=engine, if_exists='replace')
```

```python
print(pd.io.sql.get_schema(df1, name='yellow_taxi_data', con=engine))
```

```python
%time
df1.to_sql(name='yellow_taxi_data', con=engine, if_exists='append')
```
