```python
df1.tpep_pickup_datetime = pd.to_datetime(df1.tpep_pickup_datetime)
df1.tpep_dropoff_datetime = pd.to_datetime(df1.tpep_dropoff_datetime)
```

```python
df_iter = pd.read_csv('nyt.csv', iterator=True, chunksize=100000)
```

```python
type(df_iter)
```

```python
df = next(df_iter)
len(df)
```

```python
df2 = df.copy()
```

```python
df2 = df2 + df.copy()
```

```python
len(df2)
```

```python
print(pd.io.sql.get_schema(df, name='yellow_taxi_data', con=engine))
```

```python
df.tpep_pickup_datetime = pd.to_datatime(df.tpep_pickup_datetime)
df.tpep_dropoff_datetime = pd.to_datatime(df.tpep_dropoff_datetime)
```

```python
df.head(n=0).to_sql(name='yellow_taxi_data', con=engine, if_exists='replace')
```

```python
try:
    while True:
        t_start = time()

        try:
            df = next(df_iter)
        except StopIteration:
            print('No more dato to process')
            break
            
        df.tpep_pickup_datetime = pd.to_datatime(df.tpep_pickup_datetime)
        df.tpep_dropoff_datetime = pd.to_datatime(df.tpep_dropoff_datetime)

        try:
            df.to_sql(name='yellow_taxi_data', con=engine, if_exists='append')
        except Exception as e:
            print(f'Error inserting chunk into database: {e}')
            continue
        t_end = time()

        print(f'Inserted another chunk, took {} seconds')


except Exception as e:
    print(f'An error occurred during processing: {e}')
```
